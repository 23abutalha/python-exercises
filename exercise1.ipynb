```python
def string_ops(s):
    reversed_str = s[::-1]
    every_second_char = s[::2]
    return reversed_str, every_second_char
```

```python
def squares_of_evens(numbers):
    return [x**2 for x in numbers if x % 2 == 0]
```

```python
def char_frequency(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq
```

```python
def rotate_list(lst, k):
    k = k % len(lst) if lst else 0 # Handle k > list length
    return lst[-k:] + lst[:-k]
```

```python
def odd_indices(lst):
    return [i for i, v in enumerate(lst) if v % 2 != 0]
```

```python
def lists_to_dict(keys, values):
    return dict(zip(keys, values))
```

```python
def manual_iterate(lst):
    it = iter(lst)
    while True:
        try:
            print(next(it))
        except StopIteration:
            print("Iteration complete.")
            break
```

```python
def flatten_list(nested):
    return [item for sublist in nested for item in sublist]
```

```python
def is_palindrome(s):
    return s == s[::-1]
```

```python
def second_largest(numbers):
    if len(set(numbers)) < 2: return None
    nums = list(set(numbers)) # Remove duplicates
    nums.remove(max(nums))
    return max(nums)
```
